```python
import pandas as pd
from multiplex_model.clinical import LabelEncoder

m = pd.read_csv('../melted_table/results/melted_table.csv')  # or MELTED_TABLE_PATH
subset = 'danenberg'   # replace with your subset
feature = 'Grade'      # replace with your cli_feat_for_subset
df = m[(m['dataset']==subset) & (m['feature']==feature)]
print('Total rows for subset/feature:', len(df))
print('Any NaNs in value?', df['value'].isna().any())
print(df['value'].value_counts())
classes = df.dropna(subset=['value'])['value'].unique()
print('Classes deduped:', classes)
le = LabelEncoder(classes)
print('Encoder dict:', le.get_dict())
majority_frac = df['value'].value_counts(normalize=True).max()
print('Majority-class baseline F1 approx (macro-F1 baseline):', majority_frac)
```
